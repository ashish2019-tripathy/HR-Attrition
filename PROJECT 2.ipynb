## EDA 

##### Univariate Analysis -

```python
sns.countplot(x=df.Department)
```

```python
sns.countplot(x=df.EducationField)
```

```python
cat_col=[]

for i in df:
    if len(df[i].unique())<=20:
        cat_col.append(i)
```

```python
cat_col
```

```python
len(cat_col)
```

```python
con_col=[]

for i in df:
    if len(df[i].unique())>=20:
        con_col.append(i)
```

```python
con_col
```

```python
len(con_col)
```

```python
df_cat=df[cat_col]
df_cat.head()
```

```python
df_con=df[con_col]
df_con.head()
```

```python
plt.figure(figsize = (10,30))
plotnumber = 1
for i in df_cat:
    plt.subplot(13,2, plotnumber)
    sns.countplot(x = df_cat[i])
    plt.xticks(rotation = 20)
    plotnumber = plotnumber + 1
plt.tight_layout()
```

```python
plt.figure(figsize = (10,30))
plotnumber = 1
for i in df_con:
    plt.subplot(13,2, plotnumber)
    sns.countplot(x = df_con[i])
    plt.xticks(rotation = 20)
    plotnumber = plotnumber + 1
plt.tight_layout()
```

#### Bivariate Analysis-

```python
plt.figure(figsize=(10,30))
plotnumber=1
for i in df_cat:
    plt.subplot(14,2, plotnumber)
    sns.countplot(x=df_cat[i], hue=df['Attrition'])
    plotnumber=plotnumber+1
plt.tight_layout()
```
